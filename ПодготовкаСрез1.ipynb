# panel-lags-and-diff.ipynb

```python
import pandas as pd

from linearmodels import PooledOLS          # Pooled model
from linearmodels import RandomEffects      # Random-effect model
from linearmodels import PanelOLS           # Fixed-effect model
from linearmodels import FirstDifferenceOLS # First difference model

from linearmodels.panel import compare # Compare the results of multiple models
```

```python
# Загрузим данные
df = pd.read_csv('Wages.csv')
df.head(20)
```

```python
# Зададим структуру панельных данных
panel_df = df.set_index(['id', 'time']) #порядок важен, сначала всегда индив индекс
panel_df.head(20)
```

```python
panel_df['d_wks'] = panel_df.groupby(level=0)['wks'].diff() # добав новую переменную d_wks. Level=0 - индив индекс (1 столбец)
panel_df.head(20)
```

```python
panel_df['d2_wks'] = panel_df.groupby(level=0)['wks'].diff().diff() # добав новую переменную d2_wks. Level=0 - индив индекс (1 столбец)
panel_df.head(20)
```

```python
panel_df['lag_wks'] = panel_df.groupby(level=0)['wks'].shift() # добавлеям лаговую переменную (например, хотим посмотреть, как на сегодня влияет вчера)
panel_df.head(20)
```

# baseline-regression.ipynb

```python
import numpy as np
import pandas as pd

from linearmodels import PooledOLS          # Pooled model
from linearmodels import RandomEffects      # Random-effect model
from linearmodels import PanelOLS           # Fixed-effect model
from linearmodels import FirstDifferenceOLS # First difference model

from linearmodels.panel import compare      # Compare the results of multiple models
from statsmodels.api import add_constant    # for matrices of regression design
```

```python
df = pd.read_csv('Gasoline.csv')
df.head()
```

```python
panel_df = df.set_index(['country', 'year']) #(Порядок индексов важен! Сначала индивидуальный, потом временной
panel_df.head()
```

## Подгонка модели через формулу

Важно в формуле обязательно нужно указать 1+ для учёта константы: endog~1+exog1+.... Иначе будет оцениваться модель без константы

## Модель Пула (OLS-оценивание)

### не учитыв индив эффекты

```python
mod_pl = PooledOLS.from_formula(formula='lgaspcar~1+lincomep+lrpmg+lcarpcap', data=panel_df)
res_pl = mod_pl.fit()
res_pl
```

Parameter - коэф модели 

## Модель Случайных эффектов/Random Effects/RE (GLS-оценивание)

### учитыв индив эффекты, но нет коррел с регрессорами

```python
mod_re = RandomEffects.from_formula(formula='lgaspcar~1+lincomep+lrpmg+lcarpcap', data=panel_df)
res_re = mod_re.fit()
res_re
```

lincomep без учета индив эфф больше чем с учетом (RE)

## Модель Фиксированных эффектов/Fixed Effects/FE (Within-оценка)

### учитыв индив эффекты и допускает их коррел с регрессорами, то есть самый общий случай

### но не оценивает переменные, которые постоянны во времени

```python
mod_fe = PanelOLS.from_formula(formula='lgaspcar~1+lincomep+lrpmg+lcarpcap+EntityEffects', data=panel_df, drop_absorbed=True)
res_fe = mod_fe.fit()
res_fe
#drop_absorbed=True для автоматического удаление постоянных во времени переменных (иначе будет сообщение об ошибке в случае наличия таковых)
```

## Модель Фиксированных эффектов/Fixed Effects/FE (FD-оценка)

### тут без 1 в модели, так как при дифф константа уходит

```python
mod_fd = FirstDifferenceOLS.from_formula(formula='lgaspcar~lincomep+lrpmg+lcarpcap', data=panel_df)
res_fd = mod_fd.fit()
res_fd
```

```python
res_pl.params.round(3) #чтобы только коэф.
```

# Панельные регрессии с робастными инференциями

```python
# Подгоним модели с оценкой ковариационной матрицы Ареллано-Бонда

mod_pl = PooledOLS.from_formula(formula='lgaspcar~1+lincomep+lrpmg+lcarpcap', data=panel_df)
mod_re = RandomEffects.from_formula(formula='lgaspcar~1+lincomep+lrpmg+lcarpcap', data=panel_df)
mod_fe = PanelOLS.from_formula(formula='lgaspcar~1+lincomep+lrpmg+lcarpcap+EntityEffects', data=panel_df)
mod_fd = FirstDifferenceOLS.from_formula(formula='lgaspcar~lincomep+lrpmg+lcarpcap', data=panel_df)

res_pl = mod_pl.fit(cov_type='clustered', cluster_entity=True) # cluster_entity - учит сер корр, но не учит кросс-корр (наличие общих шоков)
res_re = mod_re.fit(cov_type='clustered', cluster_entity=True)
res_fe = mod_fe.fit(cov_type='clustered', cluster_entity=True)
res_fd = mod_fd.fit(cov_type='clustered', cluster_entity=True)

compare({'Pool': res_pl, 'RE': res_re, 'FE': res_fe, 'FD':res_fd}, stars=True)
```

pvalue маленькая, то есть каждая регрессия отдельно значима

*** - значим на 1% (5 b 10 % тоже соответственно)

** - значим на 5%, но не значим на 1%

*- значим на 10%

```python
# Подгоним модели с оценкой ковариационной матрицы Дрисколла-Края

mod_pl = PooledOLS.from_formula(formula='lgaspcar~1+lincomep+lrpmg+lcarpcap', data=panel_df)
mod_re = RandomEffects.from_formula(formula='lgaspcar~1+lincomep+lrpmg+lcarpcap', data=panel_df)
mod_fe = PanelOLS.from_formula(formula='lgaspcar~1+lincomep+lrpmg+lcarpcap+EntityEffects', data=panel_df)
mod_fd = FirstDifferenceOLS.from_formula(formula='lgaspcar~lincomep+lrpmg+lcarpcap', data=panel_df)

res_pl = mod_pl.fit(cov_type='kernel')
res_re = mod_re.fit(cov_type='kernel')
res_fe = mod_fe.fit(cov_type='kernel')
res_fd = mod_fd.fit(cov_type='kernel')

compare({'Pool': res_pl, 'RE': res_re, 'FE': res_fe, 'FD':res_fd}, stars=True, precision='std_errors')
```

```python
# Подгоним для 1 модели но разными оценками ковар матрицы

mod = PanelOLS.from_formula(formula='lgaspcar~1+lincomep+lrpmg+lcarpcap+EntityEffects', data=panel_df)

res_ols = mod.fit()
res_hc = mod.fit(cov_type='robust')
res_ab = mod.fit(cov_type='clustered', cluster_entity=True)
res_dk = mod.fit(cov_type='kernel')

compare({'OLS': res_ols, 'HC': res_hc, 'AB': res_ab, 'DK': res_dk}, stars=True, precision='std_errors')
```

# Тестирование гипотез, F-тест

```python
# Загрузим данные
df = pd.read_csv('Wages.csv')
# Зададим структуру панельных данных
panel_df = df.set_index(['id', 'time']) #порядок важен, сначала всегда индив индекс
panel_df.head(20)
```

```python
# Модель Пула
mod_pl = PooledOLS.from_formula(formula=' lwage ~1+ ed+ exp+ I(exp**2)+ wks', data=panel_df)
res_pl = mod_pl.fit()
res_pl
```

F-статистика = 411.62, pvalue = 0 < альфа (5%) -> значима, отвергаем H0

1 степень свободы = 4, так как занулили 4 коэф

```python
res_pl.f_statistic
```

```python
res_pl.wald_test(formula='exp=0, I(exp ** 2)=0')
```

F-статистика = 723.86, pvalue = 0 < альфа (5%) -> значимо, отвергаем H0

```python
# Модель Пула (робастная матрица Ареллано-Бонда)
mod_pl = PooledOLS.from_formula(formula=' lwage ~1+ ed+ exp+ I(exp**2)+ wks', data=panel_df)
res_pl = mod_pl.fit(cov_type='clustered', cluster_entity=True)
res_pl
```

```python
res_pl.f_statistic_robust
```

F-статистика = 72.68, pvalue = 0 < альфа (5%) -> значима, отвергаем H0

```python
res_pl.wald_test(formula='exp=0, I(exp ** 2)=0')
```

F-статистика = 166.6, pvalue = 0 < альфа (5%) -> значимо, отвергаем H0

Аналогично с моделями RE и FE:

* если неробастный F-тест, то подгоняем модель и res_pl.f_statistic
* если неробастный F-тест на совместную знач-ть нескольких регрессоров, то подгоняем модель и res_pl.wald_test(formula='exp=0, I(exp ** 2)=0')
* если робастный F-тест, то подгоняем модель с оценкой ковариационной матрицы Ареллано-Бонда и res_pl.f_statistic_robust
* если робастный F-тест на совместную знач-ть нескольких регрессоров, то подгоняем модель с оценкой ковариационной матрицы Ареллано-Бонда и res_pl.wald_test(formula='exp=0, I(exp ** 2)=0')
